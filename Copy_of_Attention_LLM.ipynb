```python
# Attention Formula
Attention(Q, K, V) = softmax(Q X k ^ T / sqrt(d_k)) X V
```



*   Q is the Query matrix ( compare what each word is looking for)

*   K is the Key matrix (what every other word offers)

*   V is the value matrix ( using those omparisons to collect the actual info from the most relevant words
*   K^T is the transpose of the Key matrix


*   d_k is the dimension of the Key vector


*   sqrt means square root



# Setting Up: From Words to Vector

```python
# A simple sentence with 3 words: "I love AI"
# Transformer converts each word into a vector (a list of numbers), called an embedding. Let's use d_emb = 4

word_embeddings = {
    "I": [1, 0, 1, 0],
    "love": [0, 1, 0, 1],
    "AI": [1, 1, 0, 0]
}

# You can access them like this:
# print(word_embeddings["I"])
# print(word_embeddings["love"])
# print(word_embeddings["AI"])
```

```python
import numpy as np

# Stack these into an input matrix X of shape 3 x 4 (3 words, each with 4 numbers)

X = np.array([
    [1, 0 , 1, 0],
    [0, 1, 0, 1],
    [1, 1, 0, 0]
])
# here each row is one word and each column is one number in the embedding

print(X)
```

# Creating Q, K, and V matrices

## Multiply the input matrix X with three seqarate weight metrices: W_Q, W_K, and W_V. These matrices are learned during training.

```python
import numpy as np

# Assuming d_emb = 4 (from previous cells) and we want d_model = 4 for Q, K, V
d_model = 4

# Initialize W_Q, W_K, W_V as random weight matrices
W_Q = np.random.rand(X.shape[1], d_model) # (d_emb, d_model)
W_K = np.random.rand(X.shape[1], d_model) # (d_emb, d_model)
W_V = np.random.rand(X.shape[1], d_model) # (d_emb, d_model)

# Calculate Q, K, V using matrix multiplication
Q = X @ W_Q
K = X @ W_K
V = X @ W_V

print("Q matrix:\n", Q)
print("K matrix:\n", K)
print("V matrix:\n", V)
```

```python
# Calculate unscaled attention scores
attention_scores = Q @ K.T

print("Attention Scores (Unscaled):\n", attention_scores)
```

# Scaling by dividing every scores by sqrt(d_k)

Why? If the dot product values are too large, the softmax function in the next step will produce very extreme values (0 or 1), this makes it difficult for the model to learn. Scaling keeps the score values in a managable range

```python
Scaled_scores = attention_scores / np.sqrt(d_model)

print("Scaled Attention Scores:\n", Scaled_scores)
```

# Applying Softmax to each row - softmax converts the scores into probabilities that add up to 1.

```python
# Calculate e^(x) for all scaled scores
exp_scores = np.exp(Scaled_scores)

# Divide by the sum of e^(x) for each row (axis=1 means row-wise)
# keepdims=True ensures the dimensions align for division
attention_weights = exp_scores / np.sum(exp_scores, axis=1, keepdims=True)

print("Attention Weights (Probabilities):\n", attention_weights)

# Let's verify that each row adds up to 1
print("\nSum of each row:", np.sum(attention_weights, axis=1))
```

# Computing the Final Output (Attention Weights X V)

```python
output = attention_weights @ V

print("Final Output (Attention Weights @ V):\n", output)
```
